```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

#Paramètres
Theta = 0.6
eps = 0.1
bruit = 0.2
m_list = [5, 10, 20, 50, 100, 200]
reps = 500
n_test = 20_000
```

```python

# Générer les données

def dataset(m, theta=Theta, noise=0.0):
    X = np.random.rand(m)
    y = np.zeros(m)
    for i in range(m):
        if X[i] > theta:
            y[i] = 1
        else:
            y[i] = 0

    # Ajout du bruit si nécessaire
    if noise > 0:
        for i in range(m):
            if np.random.rand() < noise:
                y[i] = 1 - y[i]

    return X, y
```

```python
# Visualisation d échantillon
X_demo, y_demo = dataset(50, noise=0.1)

plt.figure(figsize=(7, 4))
plt.plot(X_demo[y_demo == 0], np.zeros(sum(y_demo == 0)), 'o', color='BLUE', label='Classe 0')
plt.plot(X_demo[y_demo == 1], np.ones(sum(y_demo == 1)), 'o', color='GREEN', label='Classe 1')
plt.axvline(Theta, color='BLACK', linestyle='--', label='Seuil theta')
plt.title("Echantillon avec du bruit ")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```

L'image montre l'impact du bruit avec des points mal classés de chaque côté du seuil, illustrant le passage au cadre agnostique.

```python
# Algorithme ERM (Minimisation du Risque Empirique)
def predict_threshold(x, threshold):
    return (np.asarray(x) > threshold).astype(int)

def erm_threshold(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=int)
    if len(x) == 0:
        raise ValueError("L'échantillon d'entraînement ne doit pas être vide.")

    xs = np.sort(np.unique(x))
    candidates = np.concatenate(([0.0], xs, [1.0], (xs[:-1] + xs[1:]) / 2.0))
    candidates = np.unique(candidates)

    errors = np.array([np.mean(predict_threshold(x, t) != y) for t in candidates])
    best_idx = int(np.argmin(errors))
    return float(candidates[best_idx]), float(errors[best_idx])
```

```python
#  Test de l'ERM
t_demo, e_demo = erm_threshold(X_demo, y_demo)
print(f"Seuil estimé par ERM : {t_demo:.3f}")
print(f"Erreur empirique minimale : {e_demo:.3f}")
```

```python
#  Définition de la fonction cible
def target_function(x, theta=Theta):
    """Retourne les étiquettes cibles : 1 si x > theta, sinon 0."""
    return (np.asarray(x) > theta).astype(int)
```

```python

#   Évaluation des erreurs (Empirique vs Test)
def evaluate_model(threshold, theta=Theta, noise_rate=0.0):
    """Estime l’erreur de test sur un grand échantillon indépendant."""
    x_test = np.random.uniform(0.0, 1.0, size=n_test)
    y_test = target_function(x_test, theta)
    if noise_rate > 0:
        flips = np.random.rand(n_test) < noise_rate
        y_test = np.where(flips, 1 - y_test, y_test)
    return float(np.mean(predict_threshold(x_test, threshold) != y_test))
```

```python
# Comparaison pour le cadre réalisable (bruit = 0.0)
rows = []
for m in m_list:
    x, y = dataset(m, noise=0.0)
    threshold, empirical_error = erm_threshold(x, y)
    true_error_est = evaluate_model(threshold, noise_rate=0.0)
    rows.append((m, threshold, empirical_error, true_error_est))

print("\n--- Résultats du Cadre Réalisable ---")
print(f"{'m':>5} {'threshold':>12} {'empirical error':>18} {'test error':>13}")
for row in rows:
    print(f"{row[0]:5d} {row[1]:12.3f} {row[2]:18.3f} {row[3]:13.4f}")
```

```python
# Graphe Erreur empirique vs Erreur de test
ms = [r[0] for r in rows]
emp = [r[2] for r in rows]
test = [r[3] for r in rows]

plt.figure(figsize=(8, 4))
plt.plot(ms, emp, "o-", label="Erreur empirique")
plt.plot(ms, test, "o-", label="Erreur de test indépendante")
plt.xscale("log")
plt.xlabel("Taille de l’échantillon m ")
plt.ylabel("Taux d’erreur")
plt.title("Cadre Réalisable : Erreur empirique vs Erreur de test")
plt.grid()
plt.legend()
plt.show()
```

 Ce graphique compare l'évolution du taux d'erreur empirique et de l'erreur de test indépendante en fonction de la taille de l'échantillon $m$.  L'erreur empirique reste constamment à zéro, tandis que l'erreur de test indépendante diminue progressivement à mesure que la taille de l'échantillon $m$ augmente pour converger également vers zéro.

---


 Ce graphique valide visuellement le concept de convergence statistique : l'écart entre l'erreur empirique (toujours optimiste à 0) et l'erreur réelle (sur le test) se résorbe à mesure que l'information disponible ($m$) grandit. C'est la base mathématique qui permet de garantir qu'un modèle apprend « valablement » à partir d'exemples empiriques.

**Conclusion**
Les expériences illustrent la différence essentielle entre les deux cadres. L’apprentissage PAC suppose que le problème est réalisable : une hypothèse de la classe représente parfaitement la règle d’étiquetage. L’ERM peut alors obtenir une erreur empirique nulle, tandis que son erreur réelle diminue généralement lorsque la quantité de données augmente. L’apprentissage PAC agnostique ne suppose pas qu’une hypothèse représente parfaitement la distribution des étiquettes observées. En présence de bruit, l’ERM minimise l’erreur empirique, mais ne peut généralement pas éliminer l’erreur en population. L’objectif est de se rapprocher du meilleur risque atteignable dans la classe d’hypothèses. Les graphiques et tableaux du notebook permettent d’observer expérimentalement ces différences.
